## 1.series

### 1.1 series的创建

```python
import pandas as pd
import numpy as np
s = pd.Series([10,2,3,4,5,6])
print(s)
# 自定义索引
s = pd.Series([10,2,3,4,5,6],index=['a','b','c','d','e','f'])
print(s)
# 定义name
s = pd.Series([10,2,3,4,5],index=['a','b','c','d','e'],name='月份')
print(s)
```

```python
# 通过字典来创建
s = pd.Series({'a':1,'b':2,'c':3})
print(s)
s1 = pd.Series(s,index=['a','b'])
print(s1)
```

### 1.2 series的属性

```python
# series的属性
print(s.index)
print(s.values)
print(s.shape,s.ndim,s.size,s.dtypes)
s.name = 'qwe'
print(s.name)
print(s.loc['a'])   # 显示索引
print(s.iloc[1])    # 隐式索引
print(s.at['a'])
print(s.iat[1])
```

### 1.3访问数据

```python
print(s['a'])
print(s[s<3])
s['d'] = 4
print(s.head())
print(s.tail(1))
```

### 1.4series常用方法

```python
s = pd.Series([10,2,np.nan,None,3,4,5],index = ['A','B','C','D','E','F','G'],name = 'data')
print(s)
```

```python
s.head(3).to_frame() # 默认取前5行的数据
s.tail().to_frame()  # 默认取后5行的数据
```

```python
# 查看所有的描述性信息
s.describe().to_frame()
```

```python
# 获取元素个数(忽略缺失值)
print(s.count())
```

```python
# 获取索引
print(s.keys()) # 方法
print(s.index)  # 属性
```

```python
s.isna()    # 检查Serier里面的每一个元素是否缺失
```

```python
s.isin([4,5,6]) # 检查
```

```python
s.describe()
```

```python
print(s.mean()) # 平均值
print(s.std())  # 标准差
print(s.sum())  # 总和
print(s.var())  # 方差
print(s.min())  # 最小值
print(s.max())  # 最大值
print(s.median()) #中位数
```

```python
print(s.sort_values().to_frame())
print(s.quantile(0.8))
```

```python
# 众数
s['H'] = 4
print(s.mode())
```

```python
print(s.value_counts()) # 每个元素的计数
```

```python
print(s.drop_duplicates()) # 去重返回list
print(s.unique())  # 去重返回array
print(s.nunique()) # 去重后的元素个数
```

```python
# 排序    值、索引
print(s.sort_index())   # 按索引排序
print(s.sort_values())  # 按值排序
```
